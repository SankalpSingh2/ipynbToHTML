```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider

# Define the initial parameters
x = np.linspace(0, 10, 1000)
initial_freq = 1

# Create the figure and axis
fig, ax = plt.subplots()
plt.subplots_adjust(bottom=0.2)  # Adjust layout for slider

# Plot the initial sine wave
[line] = ax.plot(x, np.sin(2 * np.pi * initial_freq * x), label=f"Frequency: {initial_freq} Hz")

# Set labels and title
ax.set_xlabel("Time")
ax.set_ylabel("Amplitude")
ax.set_title("Interactive Sine Wave")
ax.legend()

# Add a slider for frequency adjustment
ax_slider = plt.axes([0.2, 0.05, 0.65, 0.03])  # x, y, width, height
slider = Slider(ax_slider, "Frequency", 1, 50, valinit=initial_freq, valstep=1)

# Update function for slider
def update(freq):
    line.set_ydata(np.sin(2 * np.pi * freq * x))
    line.set_label(f"Frequency: {int(freq)} Hz")
    ax.legend()
    fig.canvas.draw_idle()

# Connect the slider to the update function
slider.on_changed(update)

plt.show()
```
