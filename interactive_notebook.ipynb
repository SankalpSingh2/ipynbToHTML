```python
import plotly.graph_objects as go
import numpy as np
import plotly.offline as pyo

# Initialize Plotly for offline use
pyo.init_notebook_mode(connected=True)

# Create figure
fig = go.Figure()
x = np.linspace(0, 10, 1000)
initial_freq = 1
fig.add_trace(go.Scatter(x=x, y=np.sin(2 * np.pi * initial_freq * x)))

# Create slider
steps = []
for freq in range(1, 51):
    step = dict(
        method="restyle",
        args=[{"y": [np.sin(2 * np.pi * freq * x)]}],
        label=f"{freq} Hz"
    )
    steps.append(step)

sliders = [dict(
    active=0,
    currentvalue={"prefix": "Frequency: "},
    steps=steps
)]

fig.update_layout(
    sliders=sliders,
    title="Interactive Sine Wave",
    height=600
)

# Save the figure as an interactive HTML div
plot_html = pyo.plot(fig, include_plotlyjs=True, output_type="div")

# Save to a file so it can be injected into HTML later
with open("failed_attempts/plotly_figure.html", "w", encoding="utf-8") as f:
    f.write(plot_html)
```
