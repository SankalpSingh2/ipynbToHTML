```python
from bokeh.embed import components
from bokeh.plotting import figure
from bokeh.models import Button, CustomJS
from bokeh.layouts import column
from bokeh.io import output_notebook
from IPython.display import HTML

output_notebook()

# plot creation
p = figure(width=400, height=300, title="Interactive Demo")
line = p.line([1, 2, 3], [4, 1, 2], line_width=2, color="blue", name="line")
scatter = p.scatter([1, 2, 3], [4, 1, 2], size=10, color="red", name="scatter")

# buttons with js callbacks
button_line = Button(label="Show/Hide Line")
button_scatter = Button(label="Scatter/Line")

# js callback for line button
callback_line = CustomJS(args=dict(line=line), code="""
    line.visible = !line.visible
""")
button_line.js_on_click(callback_line)

# js callback for the scatter button
callback_scatter = CustomJS(args=dict(scatter=scatter, line=line), code="""
    if (scatter.visible) {
        scatter.visible = false;
        line.visible = true;
    } else {
        scatter.visible = true;
        line.visible = false;
    }
""")
button_scatter.js_on_click(callback_scatter)

layout = column(p, button_line, button_scatter)

# convert bokeh layout to html components
script, div = components(layout)

# embed bokeh plot, using js callbacks instead of pythojn for browser side manipulation
html_content = f"""
<!DOCTYPE html>
<html>
<head>
<link
    href="http://cdn.bokeh.org/bokeh/release/3.4.0/bokeh.min.css"
    rel="stylesheet"
    type="text/css"
>
<link
    href="http://cdn.bokeh.org/bokeh/release/3.4.0/bokeh-tables.min.css"
    rel="stylesheet"
    type="text/css"
>
<link
    href="http://cdn.bokeh.org/bokeh/release/3.4.0/bokeh-widgets.min.css"
    rel="stylesheet"
    type="text/css"
>

<script src="http://cdn.bokeh.org/bokeh/release/3.4.0/bokeh.min.js"></script>
<script src="http://cdn.bokeh.org/bokeh/release/3.4.0/bokeh-tables.min.js"></script>
<script src="http://cdn.bokeh.org/bokeh/release/3.4.0/bokeh-widgets.min.js"></script>
</head>
<body>
    {script}
    {div}
</body>
</html>
"""

# Display the HTML content
HTML(html_content)
```
